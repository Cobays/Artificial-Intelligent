# Clean and Split Chinese-English Dataset
This notebook reads a parallel corpus (Chinese and English), cleans the text, and outputs aligned sentences to `source.txt` and `target.txt`.

```python
import os
import re

# Set base directory
base_dir = r"C:\\Users\\Kevin\\Desktop\\Artificial Intelligent\\Artificial-Intelligent"

# Input file paths
input_chinese = os.path.join(base_dir, "data", "archive", "chinese.zh")
input_english = os.path.join(base_dir, "data", "archive", "english.en")

# Output file paths
output_clean_zh = os.path.join(base_dir, "data", "source.txt")
output_clean_en = os.path.join(base_dir, "data", "target.txt")
```

```python
def clean_text(text, lang="en"):
    text = text.strip()
    text = re.sub(r"\s+", " ", text)
    if lang == "en":
        text = re.sub(r"[^a-zA-Z0-9\s.,!?']+", '', text)
    return text
```

```python
with open(input_chinese, "r", encoding="utf-8") as zh_file, \
     open(input_english, "r", encoding="utf-8") as en_file, \
     open(output_clean_zh, "w", encoding="utf-8") as out_zh, \
     open(output_clean_en, "w", encoding="utf-8") as out_en:

    for zh_line, en_line in zip(zh_file, en_file):
        clean_zh = clean_text(zh_line, lang="zh")
        clean_en = clean_text(en_line, lang="en")

        if clean_zh and clean_en:
            out_zh.write(clean_zh + "\n")
            out_en.write(clean_en + "\n")

print("✅ Done cleaning. Saved to source.txt and target.txt.")
```
